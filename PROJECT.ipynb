#1.2-25th, 50th, and 75th Percentiles

```python
Q1 = np.percentile(math_scores, 25)
Q2 = np.percentile(math_scores, 50)
Q3 = np.percentile(math_scores, 75)

IQR = Q3 - Q1

print(Q1)
print(Q2)
print(Q3)
print(IQR)
print(df)
```

#1.3-Outliers Count (IQR Rule)

```python
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = math_scores[(math_scores < lower) | (math_scores > upper)]

print(len(outliers))
print(outliers)
print(df)
```

#2.2-Handle Missing Values
Columns with more than 40% missing values would be removed if present. Any remaining missing values in numeric columns were filled using the median. In this dataset, no significant missing values were found.

```python
threshold = len(df) * 0.4
df = df.dropna(axis=1, thresh=len(df) - threshold)
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

print(df.isnull().sum())
```

#2.3-Remove Duplicate Rows

```python
df = df.drop_duplicates()
print(df.duplicated().sum())
```

#2.4-Create at least New Derived Column

```python
df["Average Score"] = (
    df["math score"] +
    df["reading score"] +
    df["writing score"]
) / 3

df.head()
```

```python
gender_group = df.groupby("gender")[["math score"]].sum()

print(gender_group)
```

```python
gender_group = df.groupby("gender")[["reading score", "writing score"]].count()

print(gender_group)
```

#(3)MATPLOTLIB

#3.1-LINE CHART

```python
data = df.groupby("gender")["math score"].mean()

plt.figure(figsize=(6,4))
plt.plot(data.index, data.values, marker='o')
plt.title(" Math Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Math Score")
plt.grid(True)
plt.show()
```

#3.2-Bar Chart

```python
data = df.groupby("gender")["Average Score"].mean()
colors = ["blue", "green"]
plt.figure(figsize=(6,4))
plt.bar(data.index,data.values, color=colors)
plt.title("Average Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Score")
plt.show()
```

#3.3-Histogram

```python
plt.figure(figsize=(7,4))

plt.hist(df["math score"], bins=15, edgecolor="black")

plt.axvline(color="red", linestyle="--", label="Lower Bound")
plt.axvline(color="green", linestyle="--", label="Upper Bound")

plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Frequency")
plt.legend()

plt.show()
```

#3.4-Scatter Plot

```python
plt.figure(figsize=(6,5))

for gender in df["gender"].unique():
    subset = df[df["gender"] == gender]
    plt.scatter(
        subset["math score"],
        subset["reading score"],
        label=gender
    )
    plt.title("Math Score vs Reading Score")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")
plt.legend()
plt.show()
```

#3.5-2×2 Dashboard

```python
fig, axs = plt.subplots(2, 2, figsize=(12,10))

# Line Chart
axs[0,0].plot(data.index,data.values, marker='o')
axs[0,0].set_title("Average Math Score by Gender")

# Bar Chart
axs[0,1].bar(data.index,data.values)
axs[0,1].set_title("Average Score by Gender")

# Histogram
axs[1,0].hist(df["math score"], bins=15, edgecolor="black")
axs[1,0].axvline(lower, linestyle="--")
axs[1,0].axvline(upper, linestyle="--")
axs[1,0].set_title("Math Score Distribution")

# Scatter Plot
for gender in df["gender"].unique():
    subset = df[df["gender"] == gender]
    axs[1,1].scatter(
        subset["math score"],
        subset["reading score"],
        label=gender
    )

axs[1,1].set_title("Math vs Reading")
axs[1,1].legend()

plt.tight_layout()

plt.savefig("dashboard.png")
plt.show()
```
